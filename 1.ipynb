```python
## OpenRouter示例代码 免费
from openai import OpenAI

client = OpenAI(
  base_url="https://openrouter.ai/api/v1",
  api_key=openai.api_key,
)

completion = client.chat.completions.create(
  extra_headers={
    # "HTTP-Referer": "<YOUR_SITE_URL>", # Optional. Site URL for rankings on openrouter.ai.
    # "X-Title": "<YOUR_SITE_NAME>", # Optional. Site title for rankings on openrouter.ai.
  },
  extra_body={},
  # model="deepseek/deepseek-v3-base:free", # 这个 "tokens_completion": 0 没有答案
  model="deepseek/deepseek-chat-v3-0324:free",
  messages=[
    {
      "role": "user",
      "content": "What is the meaning of life? write a haiku about it."
    }
  ]
)
print(completion.choices[0].message.content)
# 1. 报错如下，解决方法为 https://openrouter.ai/settings/privacy 中 Model Training 开启
# NotFoundError: Error code: 404 - {'error': {'message': 'No endpoints found matching your data policy. 
# Enable prompt training here: https://openrouter.ai/settings/privacy', 'code': 404}}
# 2. 无报错但无内容，model="deepseek/deepseek-v3-base:free", # 这个 "tokens_completion": 0 没有答案
# 解决方法：更换model为 model="deepseek/deepseek-chat-v3-0324:free",
# 3. 得到答案
# **The Meaning of Life:**  
# Seeking, lost, then found—  
# love blooms in fleeting moments,  
# joy in simple things.
```
